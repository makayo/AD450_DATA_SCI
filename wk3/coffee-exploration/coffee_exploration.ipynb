- **Production vs. exports:**

## Part 4 – Data Quality Assessment

```python
# Count duplicate rows in each dataset
display("Coffee exports:", export_df.duplicated().sum())
display("Coffee imports:", import_df.duplicated().sum())
display("Coffee production:", production_df.duplicated().sum())
```

```python
# Check for inconsistent formatting and unexpected values
display("Import countries with extra spaces:", (import_df["Country"] != import_df["Country"].str.strip()).sum())
display("Negative export values:", (export_df.select_dtypes("number") < 0).sum().sum())
display("Zero values in imports:", (import_df.select_dtypes("number") == 0).sum().sum())
display("Coffee types:", production_df["Coffee type"].unique())
```

### Data Quality Findings

**Coffee exports:**
- **Missing values / duplicates:** None.
- **Negative values:** Brazil has -2,147,483,648 in 2014, 2015, and 2019 (3 values). Exports can't be negative.
  **Impact:** Brazil, the largest exporter, would look like it exported less than nothing, and averages and totals for those years would be far too low.
- **Data types:** Year columns are int64, while production uses float64.
  **Impact:** Combining or comparing datasets may need type conversion first.

**Coffee imports:**
- **Missing values / duplicates:** None.
- **Extra spaces:** 28 country names start with spaces (e.g. `'   Austria'`).
  **Impact:** Searches and joins on country names would fail, so these countries would be silently dropped when combining datasets.
- **Zeros used as missing data:** 54 zero values. For example, Belgium shows 0 in early years and Belgium/Luxembourg shows 0 in later years, because the countries were reported differently over time.
  **Impact:** Zeros would be treated as "no imports," pulling averages down and creating false drops or jumps in trends.

**Coffee production:**
- **Missing values / duplicates:** None.
- **Inconsistent coffee types:** `Robusta/Arabica` and `Arabica/Robusta` likely mean the same thing.
  **Impact:** Grouping by coffee type would split one category into two and undercount both.
- **Crop years vs. calendar years:** Production uses `1990/91`, while exports and imports use `1990`.
  **Impact:** Year columns won't line up, so comparing production to exports would need the years matched first.

## Part 5 – Initial Insights

The coffee production dataset appears most complete because it has no missing values, duplicates, or negative values. The coffee imports dataset needs the most cleaning because it has extra spaces in 28 country names and 54 zeros that likely mean missing data. A global coffee company could use these datasets to identify major producers, compare imports and exports, track trends, and find potential suppliers and new markets.
