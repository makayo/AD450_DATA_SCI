# JavaBeans Coffee Sales Analysis
## DataFrame Creation, Manipulation, and Business Insights

This notebook analyzes JavaBeans beverage sales for 2021 and 2022 using Pandas.

```python
import pandas as pd

coffee_sales = pd.DataFrame(
    {
        "Americano": [5628, 6239],
        "Latte": [12426, 25384],
        "Espresso": [852, 756],
        "Cappuccino": [8522, 7561],
    },
    index=["2021 Sales", "2022 Sales"]
)
```

```python
total_sales = coffee_sales.sum(axis=0)

print("Total Sales by Beverage (2021 to 2022):")
display(total_sales.to_frame(name="Total Sales"))
```

```python
yoy_growth = (
    (coffee_sales.loc["2022 Sales"] - coffee_sales.loc["2021 Sales"])
    / coffee_sales.loc["2021 Sales"]
) * 100

yoy_growth = yoy_growth.round(2)

print("Year-over-Year Growth (2021 to 2022):")
display(yoy_growth.to_frame(name="YoY Growth (%)"))
```

```python
for year in coffee_sales.index:
    sales = coffee_sales.loc[year]

    best_seller = sales.idxmax()
    least_seller = sales.idxmin()

    print(f"{year}:")
    print(f"  Best Selling beverage: {best_seller} ({sales.max():,} sales)")
    print(f"  Least Selling beverage: {least_seller} ({sales.min():,} sales)")
    print()
```
